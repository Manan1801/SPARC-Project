```python
# Jupyter Notebook in VS Code
# ---------------------------------------------------------
# This code reads from precomputed CSV files that contain
# columns for frames & 2D landmark coordinates.
#
# It computes distance-from-previous-frame & distance-from-first-frame
# for 'landmark #15' (left_wrist) and 'landmark #16' (right_wrist).
#
# Then it bins the distances by time interval (group_interval)
# and produces four box-and-whisker plots for a SINGLE directory:
#    1) LM #15 => Distance from Previous
#    2) LM #15 => Distance from First
#    3) LM #16 => Distance from Previous
#    4) LM #16 => Distance from First
#
# Each plot is saved as an interactive HTML file, with the bin size
# included in the file name so you can keep track of which bin size
# was used to generate each plot.
#
# You can call this function separately for your "novice" directory
# and your "expert" directory to generate 4 plots for each.

import os
import glob
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from collections import defaultdict

############################################################
# 1) Compute Distances from a Single CSV
############################################################
def compute_distances_from_csv(csv_path, fps=30.0):
    """
    Reads a CSV file with columns at least:
        ['frame',
         'left_wrist_x', 'left_wrist_y',
         'right_wrist_x', 'right_wrist_y', ...]
    and computes, for each row/frame, the following:
      - dist_prev_15, dist_first_15  (landmark #15 => left_wrist)
      - dist_prev_16, dist_first_16  (landmark #16 => right_wrist)
    returning them along with a time array (in seconds).

    Parameters
    ----------
    csv_path : str
        Path to the CSV file.
    fps : float
        Frames per second (used to convert 'frame' index to seconds).

    Returns
    -------
    times : np.ndarray
        Time values in seconds, one per row in the CSV (sorted by 'frame').
    dist_prev_15, dist_first_15 : np.ndarray
        Distances (left_wrist) from previous frame, from first frame
    dist_prev_16, dist_first_16 : np.ndarray
        Distances (right_wrist) from previous frame, from first frame
    """
    # Load CSV
    df = pd.read_csv(csv_path)

    # Sort by 'frame' in case not already sorted
    df = df.sort_values(by='frame').reset_index(drop=True)

    # Extract relevant columns for left wrist (#15)
    x15 = df['left_wrist_x'].values
    y15 = df['left_wrist_y'].values

    # Extract relevant columns for right wrist (#16)
    x16 = df['right_wrist_x'].values
    y16 = df['right_wrist_y'].values

    # Time array: assume 'frame' is a sequential frame index
    frames = df['frame'].values
    times = frames / fps

    num_frames = len(df)

    # Prepare output arrays
    dist_prev_15  = np.full(num_frames, np.nan, dtype=float)
    dist_first_15 = np.full(num_frames, np.nan, dtype=float)
    dist_prev_16  = np.full(num_frames, np.nan, dtype=float)
    dist_first_16 = np.full(num_frames, np.nan, dtype=float)

    # ---- Landmark #15 (left_wrist) ----
    if num_frames > 0:
        dist_first_15[0] = 0.0  # distance from first frame at i=0 is 0
        x0_15, y0_15     = x15[0], y15[0]
        dist_prev_15[0]  = np.nan  # no "previous frame" for i=0

    for i in range(1, num_frames):
        dist_first_15[i] = np.sqrt((x15[i] - x0_15)**2 + (y15[i] - y0_15)**2)
        dist_prev_15[i]  = np.sqrt((x15[i] - x15[i-1])**2 + (y15[i] - y15[i-1])**2)

    # ---- Landmark #16 (right_wrist) ----
    if num_frames > 0:
        dist_first_16[0] = 0.0
        x0_16, y0_16     = x16[0], y16[0]
        dist_prev_16[0]  = np.nan

    for i in range(1, num_frames):
        dist_first_16[i] = np.sqrt((x16[i] - x0_16)**2 + (y16[i] - y0_16)**2)
        dist_prev_16[i]  = np.sqrt((x16[i] - x16[i-1])**2 + (y16[i] - y16[i-1])**2)

    return (times,
            dist_prev_15, dist_first_15,
            dist_prev_16, dist_first_16)


############################################################
# 2) Accumulate Distances from CSVs in a Directory
############################################################
def accumulate_box_data_for_directory(directory_path,
                                      which_distance="prev",
                                      which_landmark=15,
                                      group_interval=10,
                                      fps=30.0):
    """
    Goes through all CSV files in 'directory_path', computes the chosen distance:
        which_distance => 'prev' or 'first'
        which_landmark => 15 (left_wrist) or 16 (right_wrist)
    and bins the data in intervals of size `group_interval` seconds.

    dist_by_bin[ bin_index ] = list of distances

    Where bin_index = floor(t / group_interval), e.g. 0 for t in [0..group_interval).

    Parameters
    ----------
    directory_path : str
        Path to directory containing CSV files.
    which_distance : str
        "prev" => distance from previous frame
        "first" => distance from first frame
    which_landmark : int
        15 => left_wrist, 16 => right_wrist
    group_interval : float
        Width (in seconds) of each time bin.
    fps : float
        Frames per second for converting CSV 'frame' index to time (if needed).

    Returns
    -------
    dist_by_bin : dict
        Dictionary { bin_index : [list_of_distances] }
    """
    dist_by_bin = defaultdict(list)

    # Gather all CSV files
    csv_paths = sorted(glob.glob(os.path.join(directory_path, "*.csv")))

    for csv_file in csv_paths:
        (times,
         dist_prev_15, dist_first_15,
         dist_prev_16, dist_first_16) = compute_distances_from_csv(csv_file, fps=fps)

        if len(times) == 0:
            continue  # no data in CSV

        # Pick the correct distance array
        if which_landmark == 15:
            dist_array = dist_prev_15 if which_distance == "prev" else dist_first_15
        else:
            dist_array = dist_prev_16 if which_distance == "prev" else dist_first_16

        # Bin the distances
        for i, t in enumerate(times):
            val = dist_array[i]
            if np.isnan(val):
                # skip if distance is NaN
                continue
            bin_index = int(np.floor(t / group_interval))
            dist_by_bin[bin_index].append(val)

    return dist_by_bin


############################################################
# 3) Plot a Box-and-Whisker Chart, Save as Interactive HTML
############################################################
def plot_box_whisker(dist_by_bin, group_interval=10, title="Box Plot"):
    """
    Creates a Plotly box-and-whiskers plot from a dict: { bin_index : [values...] }.
    Then also computes mean & median per bin, and adds two lines (Scatter traces)
    to connect those means and medians across bins.

    bin_index is floor(t / group_interval).
    Each bin on the x-axis is labeled "[start_sec]-[end_sec]s".
    """

    sorted_bins = sorted(dist_by_bin.keys())

    # We’ll collect box traces here.
    box_traces = []

    # We'll store stats for all bins so we can plot lines for mean & median.
    x_labels = []
    mean_vals = []
    median_vals = []

    for b in sorted_bins:
        start_sec = b * group_interval
        end_sec   = (b + 1) * group_interval
        bin_label = f"{start_sec}-{end_sec}s"

        bin_data = dist_by_bin[b]
        if len(bin_data) == 0:
            continue

        # ---------------------------
        # Add a Box trace
        # ---------------------------
        box_trace = go.Box(
            y=bin_data,
            name=bin_label,
            boxmean=True,
            boxpoints=False,     # change to 'all' or 'outliers' if desired
            width=0.8,
            showlegend=False     # Hide box trace from the legend
        )
        box_traces.append(box_trace)

        # ---------------------------
        # Compute mean & median
        # ---------------------------
        mean_val = np.mean(bin_data)
        median_val = np.median(bin_data)

        x_labels.append(bin_label)
        mean_vals.append(mean_val)
        median_vals.append(median_val)

    # ---------------------------
    # Build the figure and add box traces
    # ---------------------------
    fig = go.Figure()
    for trace in box_traces:
        fig.add_trace(trace)

    # ---------------------------
    # Add a scatter line for the MEAN
    # ---------------------------
    fig.add_trace(
        go.Scatter(
            x=x_labels,
            y=mean_vals,
            name="Mean",
            mode="lines+markers",
            line=dict(color="blue", width=1.5),
            marker=dict(size=5, symbol="circle"),
            showlegend=True   # Keep in the legend
        )
    )

    # ---------------------------
    # Add a scatter line for the MEDIAN
    # ---------------------------
    fig.add_trace(
        go.Scatter(
            x=x_labels,
            y=median_vals,
            name="Median",
            mode="lines+markers",
            line=dict(color="red", width=1.5, dash="dash"),
            marker=dict(size=5, symbol="diamond"),
            showlegend=True   # Keep in the legend
        )
    )

    # ---------------------------
    # Layout / Aesthetics
    # ---------------------------
    fig.update_layout(
        title=title,
        xaxis_title=f"Time bins (every {group_interval}s)",
        yaxis_title="Distance (Normalized 2D)",
        boxmode='group',
        boxgap=0.0
    )
    return fig


############################################################
# 4) A Function to Generate Plots for a Single Directory
############################################################
def generate_plots_for_directory(directory_path,
                                 extension=".html",
                                 group_interval=10,
                                 fps=30.0,
                                 save_plots=True,
                                 prefix=""):
    """
    Produces 4 box plots in total, reading from CSVs in the given directory:

      1) LM #15 => Distance from Previous
      2) LM #15 => Distance from First
      3) LM #16 => Distance from Previous
      4) LM #16 => Distance from First

    Usage:
      generate_plots_for_directory(
          directory_path="CSV_Files/Novices",
          group_interval=10,
          fps=30.0,
          save_plots=True,
          prefix="Novices"
      )

    Parameters
    ----------
    directory_path : str
        Directory containing .csv files for a single group (e.g. novices OR experts).
    extension : str
        Typically ".html" for interactive plots or ".png"/".jpg" for static.
        We'll default to ".html" to keep them interactive.
    group_interval : float
        Interval size in seconds for binning distances.
    fps : float
        Used to convert 'frame' to time in the CSVs.
    save_plots : bool
        If True, saves each plot as an interactive HTML file,
        with the bin size included in the filename.
    prefix : str
        If provided, will be used in the filenames to differentiate
        the group (e.g., "Novices" or "Experts").
    """

    # Helper function to systematically show and/or save the plot
    def show_and_save_plot(fig, filename_prefix):
        fig.show()
        if save_plots:
            outfile = f"{filename_prefix}_bin{group_interval}s{extension}"
            fig.write_html(outfile)
            print(f"Plot saved to: {outfile}")

    # 1) LM15 => Distance from Previous
    distprev_15 = accumulate_box_data_for_directory(
        directory_path, which_distance="prev", which_landmark=15,
        group_interval=group_interval, fps=fps
    )
    fig_prev_15 = plot_box_whisker(
        distprev_15,
        group_interval=group_interval,
        title=f"{prefix} (LM15) - Distance from Previous"
    )
    show_and_save_plot(fig_prev_15, f"{prefix}_LM15_DistFromPrev")

    # 2) LM15 => Distance from First
    distfirst_15 = accumulate_box_data_for_directory(
        directory_path, which_distance="first", which_landmark=15,
        group_interval=group_interval, fps=fps
    )
    fig_first_15 = plot_box_whisker(
        distfirst_15,
        group_interval=group_interval,
        title=f"{prefix} (LM15) - Distance from First"
    )
    show_and_save_plot(fig_first_15, f"{prefix}_LM15_DistFromFirst")

    # 3) LM16 => Distance from Previous
    distprev_16 = accumulate_box_data_for_directory(
        directory_path, which_distance="prev", which_landmark=16,
        group_interval=group_interval, fps=fps
    )
    fig_prev_16 = plot_box_whisker(
        distprev_16,
        group_interval=group_interval,
        title=f"{prefix} (LM16) - Distance from Previous"
    )
    show_and_save_plot(fig_prev_16, f"{prefix}_LM16_DistFromPrev")

    # 4) LM16 => Distance from First
    distfirst_16 = accumulate_box_data_for_directory(
        directory_path, which_distance="first", which_landmark=16,
        group_interval=group_interval, fps=fps
    )
    fig_first_16 = plot_box_whisker(
        distfirst_16,
        group_interval=group_interval,
        title=f"{prefix} (LM16) - Distance from First"
    )
    show_and_save_plot(fig_first_16, f"{prefix}_LM16_DistFromFirst")

    print("\nDone! 4 total box plots generated from CSV data in:", directory_path)
    if save_plots:
        print(f"All plots saved with bin size {group_interval}s included in filenames.")

# ----------------------------------------------------------------
# Example usage:
# You can call this function separately for novices and experts:
# ----------------------------------------------------------------
# generate_plots_for_directory(
#     directory_path="CSV_Files/Novices",
#     extension=".html",
#     group_interval=10,
#     fps=30.0,
#     save_plots=True,
#     prefix="Novices"
# )

generate_plots_for_directory(
    directory_path="CSV_Files/Experts/Landmark_CSVs",
    extension=".html",
    group_interval=10,
    fps=30.0,
    save_plots=True,
    prefix="Experts"
)
# ----------------------------------------------------------------
```
